# Project 03: 🛒 Shop Billing System

## 📌 Project Overview

The Shop Billing System is a beginner-friendly Python project that generates a customer invoice by accepting product information from the user.
The program calculates the subtotal, GST amount, and final bill automatically using basic Python operations and user input.

## 📋 Features

- Accept product name from the user.
- Accept product price.
- Accept product quantity.
- Accept GST percentage.
- Calculate subtotal.
- Calculate GST amount.
- Calculate the final bill.
- Display a formatted shop invoice.

## 🛠️ Technologies
- Python
- Jupyter Notebook
- Visual Studio Code (VS Code)

```python
# Project 03: Shop Billing System

# Indian Rupees Symbol
rupees = "\u20b9"

# Product Information
product_name = input("Enter Product Name: ")

# Product Price
price = float(input(f"Enter Product Price: {rupees} "))

# Product Quantity
quantity = int(input("Enter Quantity: "))

# GST Rate
gst_rate = float(input("Enter GST Rate (%): "))

# Total Amount Calculation
total_amount = price * quantity

# GST Calculate
gst_amount = total_amount * gst_rate

# Final Amount
final_amount = total_amount + gst_amount

# Display Invoice Report
print("\n========== SHOP INVOICE ==========")

print(f"Product Name: {product_name}")
print(f"Price: {rupees} {price:,.2f}")
print(f"Quantity: {quantity:,.0f}")

print("-" * 35)

print(f"Total Amount: {rupees} {total_amount:,.0f}")
print(f"GST Rate: {gst_rate}%")
print(f"GST Amount: {rupees} {gst_amount:,.2f}")

print("-" * 35)

print(f"Final Amount: {rupees} {final_amount:,.2f}")

print("-" * 35)
```
